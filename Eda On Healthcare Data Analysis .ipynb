# Data Visualization

# Piechart 

```python
df["Gender"].value_counts().plot(kind = "pie",autopct=" %1.1f%% ", title ="Gender Distribution")
plt.ylabel('Number of Patients')
plt.show()
```

# Histplot 

```python
sns.histplot(data=df,x="Age",bins=20, kde=True , color ="teal")
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Patients")
plt.show()
```

```python
sns.histplot(data=df,x="Billing Amount",bins=30,kde=True , color ="g")
plt.title("Billing Amount Distribution")
plt.xlabel("Billing Amount")
plt.ylabel("Frequency")
plt.show()
```

# Count plot 

```python
sns.countplot(data=df,x="Medical Condition",hue="Medical Condition")
plt.title("Patients by Medical Condition")
plt.xlabel("Medical Condition")
plt.ylabel("Number of Patients")
plt.xticks(rotation=30)
plt.show()
```

```python
sns.countplot(data=df,x="Test Results",color ="r")
plt.title("Test Results Distribution")
plt.xlabel("Test Result")
plt.ylabel("Number of Patients")
plt.show()
```

# Displot 

```python
sns.displot(data=df,x="Admission Type",hue ="Admission Type")
plt.title("Admission Type Distribution")
plt.xlabel("Admission Type")
plt.ylabel("Number of Patients")
plt.show()
```

```python
sns.displot(data=df,x="Insurance Provider",color = "r")
plt.title("Insurance Provider Distribution")
plt.xlabel("Insurance Provider")
plt.ylabel("Number of Patients")
plt.xticks(rotation=30)
plt.show()
```

# Bar chart 

```python
df["Medication"].value_counts().plot(kind ="bar",color ="m")
plt.title("Medication By Patients")
plt.xlabel("Medication")
plt.ylabel("Number of Patients")
plt.show()
```
