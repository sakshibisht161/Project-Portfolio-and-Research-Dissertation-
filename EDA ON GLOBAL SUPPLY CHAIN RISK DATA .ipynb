# Visualization

# Shipments by Transport Mode 

```python
sns.displot(data=df, x="Transport_Mode", hue="Transport_Mode")

plt.title("Shipments by Transport Mode")
plt.xlabel("Transport Mode")
plt.ylabel("Number of Shipments")
plt.xticks(rotation=45)
plt.show()
```

# Shipments by Product Category

```python
sns.displot(data=df, x="Product_Category", hue="Product_Category")

plt.title("Shipments by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Number of Shipments")
plt.xticks(rotation=45)
plt.show()
```

# Shipments by Product Category

```python
sns.countplot(data=df, x="Product_Category", hue="Product_Category")

plt.title("Shipments by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Number of Shipments")
plt.xticks(rotation=45)
plt.show()
```

# Weather Condition Distribution

```python
sns.barplot(data=df, x="Weather_Condition", hue="Weather_Condition")

plt.title("Weather Condition Distribution")
plt.xlabel("Weather Condition")
plt.ylabel("Number of Shipments")
plt.xticks(rotation=45)
plt.show()
```

# Carrier Reliability by Transport Mode

```python
sns.boxplot(data=df, x="Transport_Mode", y="Carrier_Reliability_Score" ,color = "red")

plt.title("Carrier Reliability by Transport Mode")
plt.xlabel("Transport Mode")
plt.ylabel("Carrier Reliability Score")
plt.xticks(rotation=45)
plt.show()
```

# Distance Distribution 

```python
sns.histplot(data=df, x="Distance_km", kde=True , color = "g")

plt.title("Shipment Distance Distribution")
plt.xlabel("Distance (km)")
plt.ylabel("Number of Shipments")
plt.show()
```

# Geopolitical Risk vs Lead Time 

```python
sns.scatterplot(
    data=df,
    x="Geopolitical_Risk_Score",
    y="Lead_Time_Days",
    hue="Disruption_Occurred"
)

plt.title("Geopolitical Risk vs Lead Time")
plt.xlabel("Geopolitical Risk Score")
plt.ylabel("Lead Time (Days)")
plt.show()
```

# Fuel Price vs Lead Time 

```python
sns.scatterplot(
    data=df,
    x="Fuel_Price_Index",
    y="Lead_Time_Days",
    hue="Transport_Mode"
)

plt.title("Fuel Price Index vs Lead Time")
plt.xlabel("Fuel Price Index")
plt.ylabel("Lead Time (Days)")
plt.show()
```

# Distance vs Lead Time

```python
sns.scatterplot(
    data=df,
    x="Distance_km",
    y="Lead_Time_Days",
    hue="Disruption_Occurred"
)

plt.title("Distance vs Lead Time")
plt.xlabel("Distance (km)")
plt.ylabel("Lead Time (Days)")
plt.show()
```

# Correlation Heatmap

```python
plt.figure(figsize=(12,8))

numeric_df = df.select_dtypes(include="number")

sns.heatmap(numeric_df.corr(),annot=True, cmap="coolwarm",fmt=".2f")

plt.title("Correlation Heatmap")
plt.show()
```

# Monthly Shipment Trend

```python
df["Date"] = pd.to_datetime(df["Date"])

monthly_shipments = df.groupby(
    df["Date"].dt.to_period("M")
).size()

plt.figure(figsize=(12,6))

monthly_shipments.plot(kind="line", marker="o" , color = "orange")

plt.title("Monthly Shipment Trend")
plt.xlabel("Month")
plt.ylabel("Number of Shipments")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Monthly Disruption Trend 

```python
monthly_disruptions = df.groupby(
    df["Date"].dt.to_period("M")
)["Disruption_Occurred"].sum()

plt.figure(figsize=(12,6))

monthly_disruptions.plot(kind="line", marker="o", color = "m")

plt.title("Monthly Disruption Trend")
plt.xlabel("Month")
plt.ylabel("Total Disruptions")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
