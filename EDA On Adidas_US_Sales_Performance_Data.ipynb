# Visualization

# Bar Chart 

```python
df["retailer"].value_counts().plot(kind ="bar",color ="m")
plt.title("Total Sales by retailer")
plt.xlabel("retailer")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df["product"].value_counts().plot(kind ="bar",color ="g")
plt.title("Total Sales by product")
plt.xlabel("Total Sales")
plt.ylabel("product")
plt.tight_layout()
plt.show()
```

```python
df["invoice_date"] = pd.to_datetime(df["invoice_date"])
df["Month"] = df["invoice_date"].dt.month_name()

monthly_sales = df.groupby("Month")["total_sales"].sum()

monthly_sales.plot(kind="bar", figsize=(10,6),title="Total Sales by Month")

plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

# Pie Chart 

```python
df["sales_method"].value_counts().plot(kind = "pie",autopct=" %1.1f%% ", title ="Method Sales Distribution")
plt.ylabel('Total Sales')
plt.show()
```

# Count plot 

```python
sns.countplot(data=df,x="region",color ="r")
plt.title("sales by region")
plt.xlabel("region")
plt.ylabel("Total Sales")
plt.show()
```

# Scatter plot

```python
plt.figure(figsize=(9,6))
sns.scatterplot(data=df, x="total_sales",y="operating_profit")
plt.title("Total Sales vs Operating Profit")
plt.xlabel("Total Sales")
plt.ylabel("Operating Profit")
plt.show()
```

# Heatmap 

```python
plt.figure(figsize=(8,6))
sns.heatmap(
    df[[ "price_per_unit", "units_sold","total_sales","operating_profit"]].corr(),
    annot=True,cmap="coolwarm",fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

# Displot 

```python
top_5_cities = ( df.groupby("city")["total_sales"].sum().nlargest(5).index)

top5_data = df[df["city"].isin(top_5_cities)]

sns.displot(data=top5_data, x="city",hue="city")

plt.title("Top 5 Cities Sales Distribution")
plt.xlabel("City")
plt.ylabel("Number of Sales")
plt.show()
```
