# Data Visualization 

# Total Flights by Airline

```python
plt.figure(figsize=(8,5))

sns.countplot(data=df, x='airline',color ="m")

plt.title('Total Flights by Airline')
plt.xlabel('Airline')
plt.ylabel('Total Flights')

plt.xticks(rotation=30)
plt.show()
```

# Flights by Source City

```python
plt.figure(figsize=(8,5))

sns.countplot(data=df,x='source_city' , color ='r')

plt.title('Flights by Source City')
plt.xlabel('Source City')
plt.ylabel('Number of Flights')

plt.xticks(rotation=30)

plt.show()
```

# Flights by stops

```python
plt.figure(figsize=(8,5))

sns.countplot(data=df,x='stops' , color ='brown')

plt.title('Flights by stops')
plt.xlabel('Stops')
plt.ylabel('Number of Flights')

plt.xticks(rotation=30)

plt.show()
```

# Average Price by class

```python
plt.figure(figsize=(8,5))

sns.displot(data=df, x='class',color ="g")

plt.title('Average Price by class')
plt.xlabel('class')
plt.ylabel(' Average Price')

plt.xticks(rotation=30)
plt.show()
```

# Total Flights by Departure Time

```python
plt.figure(figsize=(8,5))

sns.countplot(data=df, x='departure_time',color ="orange")

plt.title('Total Flights by Departure Time')
plt.xlabel('Departure Time')
plt.ylabel('Total Flights')

plt.xticks(rotation=30)
plt.show()
```

# Flight Duration vs Price

```python
plt.figure(figsize=(8,5))

sns.scatterplot(data=df,x='duration', y='price',color = "teal")

plt.title('Flight Duration vs Price')
plt.xlabel('Duration (Hours)')
plt.ylabel('Price')

plt.show()
```

# Days Left vs Price

```python
plt.figure(figsize=(8,5))

sns.scatterplot(data=df,x='days_left', y='price', color ="cyan")

plt.title('Days Left vs Price')
plt.xlabel('Days Left')
plt.ylabel('Price')

plt.show()
```

# Correlation Heatmap

```python
plt.figure(figsize=(8,6))

numeric_df = df.select_dtypes(include=np.number)

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap')

plt.show()
```
