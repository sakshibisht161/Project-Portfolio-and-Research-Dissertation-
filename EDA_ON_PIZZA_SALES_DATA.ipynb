# Visualization

# Bar PLot 

```python
sns.barplot(data=df, x="pizza_category", y="total_price" , color ="yellow")
plt.title =("Total Revenue by Pizza Category")
plt.xlabel=("Pizza Category")
plt.ylabel=("Total Revenue")
plt.show()
```

```python
sns.barplot(data=df, x="pizza_size", y="total_price" , color ="green")
plt.title =("Total Revenue by Pizza Size")
plt.xlabel=("Pizza Size")
plt.ylabel=("Total Revenue")
plt.show()
```

```python
top10 = df.groupby("pizza_name")["total_price"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,6))
sns.barplot(x=top10.values, y=top10.index , color ="m")
plt.title = ("Top 10 Pizzas by Sales")
plt.xlabel= ("Total Sales")
plt.ylabel= ("Pizza Name")
plt.show()
```

# Line Chart 

```python
plt.figure(figsize=(10,6))

df["hour"] = pd.to_datetime(df["order_time"].astype(str)).dt.hour

hourly_orders = df.groupby("hour").size().reset_index(name="orders")

sns.lineplot(data=hourly_orders, x="hour", y="orders", marker="o")

plt.title=("Orders by Hour")
plt.xlabel=("Hour of the Day")
plt.ylabel=("Number of Orders")
plt.xticks(range(24))
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))

df["order_date"] = pd.to_datetime(df["order_date"])

df["month"] = df["order_date"].dt.month

monthly_sales = df.groupby("month")["total_price"].sum().reset_index()

sns.lineplot(data=monthly_sales, x="month", y="total_price", marker="o")

plt.title=("Monthly Sales Trend")
plt.xlabel=("Month")
plt.ylabel=("Total Sales")
plt.xticks(range(1,13))
plt.tight_layout()
plt.show()
```

# Heat Map 

```python
plt.figure(figsize=(8,6))

corr = df[["quantity", "unit_price", "total_price"]].corr()

sns.heatmap( corr,annot=True, cmap="coolwarm",fmt=".2f")

plt.title = ("Correlation Heatmap")
plt.tight_layout()
plt.show()
```
